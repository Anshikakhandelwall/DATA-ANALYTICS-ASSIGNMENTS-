Task 1: Handling Missing Values (1 Mark)

Given Dataset:

data = {
    'Age': [25, np.nan, 30, 35, np.nan],


    'Salary': [50000, 60000, np.nan, 80000, 75000]
    
}

df = pd.DataFrame(data)

print(df)

Perform:

• Identify missing values

• Fill missing values using mean



```python
import pandas as pd
import numpy as np

# given data:

data={'Age':[25,np.nan,30,35,np.nan],
      'Salary':[5000,6000,np.nan,8000,75000]
      }

df=pd.DataFrame(data)
print("original dataset:\n",df)

# 1.identify missing values
print("\nMissing Values:\n", df.isnull())

print("\nCount of Missing Values:\n", df.isnull().sum())

# 2.fill missing value using mean

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())


print("\nDataFrame after filling missing values:\n", df)
```

Task 2: Outlier Detection (1 Mark)

Given:

df = pd.DataFrame({
    'Salary': [50000, 52000, 58000, 60000, 120000]
})

Perform:

• Plot boxplot

• Identify outlier

• Remove outlier using IQR



```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Salary': [50000, 52000, 58000, 60000, 120000]
})

print("oiginal dataset :\n",df)


# 1.plot boxplot

plt.boxplot(df['Salary'])
plt.title("Salary Boxplot")
plt.ylabel("Salary")
plt.show()



Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# 2.identify outlier

outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
print("\nOutliers:\n", outliers)

# 3 data affter removing outlier

df_clean = df[(df['Salary'] >= lower_bound) & (df['Salary'] <= upper_bound)]

print("\nData after removing outliers:\n", df_clean)


```

Task 5: Mini Case Study (1 Mark)

Dataset:

data = {
    'Age': [22, 25, 47, 52, 46],
    'Income': [15000, 18000, 65000, 72000, 69000]
}

df = pd.DataFrame(data)

Perform:

• Normalize income

• Detect any anomalies

• Interpret results


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    'Age': [22, 25, 47, 52, 46],
    'Income': [15000, 18000, 65000, 72000, 69000]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)

# Normalize income

df['Income_norm'] = (df['Income'] - df['Income'].min()) / (df['Income'].max() - df['Income'].min())
print("normalize income :\n",df)


# detect any anomalies

df['z_score'] = (df['Income'] - df['Income'].mean()) / df['Income'].std()
print("anomalies:\n",df)
```

Interpretation:
1. Income values (65000–72000) are significantly higher
2. May represent a different income group
3. No extreme outlier but clear data variation
4. Normalization scales values between 0–1
