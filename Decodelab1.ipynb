```python
# ============================================
# PROJECT 1 : DATA CLEANING & PREPARATION
# DecodeLabs Internship
# ============================================

# Step 1 : Import Libraries

import pandas as pd
import numpy as np
```

```python
# --------------------------------------------
# Step 2 : Load Dataset
# --------------------------------------------

df = pd.read_excel("/content/Copy of Dataset for Data Analytics.xlsx")

print("Dataset Loaded Successfully")
print()
```

```python
# --------------------------------------------
# Step 3 : Basic Information
# --------------------------------------------

print("First 5 Rows")
print(df.head())

print("\nShape of Dataset")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)
```

```python
# --------------------------------------------
# Step 4 : Check Missing Values
# --------------------------------------------

print("\nMissing Values in Each Column")

print(df.isnull().sum())

# Total Missing Values

print("\nTotal Missing Values")

print(df.isnull().sum().sum())
```

```python
# --------------------------------------------
# Step 5 : Handle Missing Values
# --------------------------------------------

# Fill CouponCode missing values

df['CouponCode'] = df['CouponCode'].fillna("No Coupon")

print("\nMissing Values After Filling")

print(df.isnull().sum())
```

```python
# --------------------------------------------
# Step 6 : Check Duplicate Rows
# --------------------------------------------

duplicates = df.duplicated().sum()

print("\nDuplicate Rows Found:")

print(duplicates)

# Remove Duplicates

df = df.drop_duplicates()

print("\nDuplicate Rows After Removal:")

print(df.duplicated().sum())
```

```python
# --------------------------------------------
# Step 7 : Check Duplicate Order IDs
# --------------------------------------------

duplicate_orderid = df['OrderID'].duplicated().sum()

print("\nDuplicate Order IDs:")

print(duplicate_orderid)
```

```python
# --------------------------------------------
# Step 8 : Correct Date Format
# --------------------------------------------

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

print("\nDate Column Converted Successfully")

print(df['Date'].head())
```

```python
# --------------------------------------------
# Step 9 : Remove Extra Spaces
# --------------------------------------------

text_columns = df.select_dtypes(include='object').columns

for col in text_columns:

    df[col] = df[col].str.strip()

print("\nText Columns Standardized")
```

```python
# --------------------------------------------
# Step 10 : Check Data Types Again
# --------------------------------------------

print("\nUpdated Data Types")

print(df.dtypes)

```

```python
# --------------------------------------------
# Step 11 : Verify No Missing Values
# --------------------------------------------

print("\nFinal Missing Values")

print(df.isnull().sum())
```

```python
# --------------------------------------------
# Step 12 : Verify Duplicate Rows
# --------------------------------------------

print("\nFinal Duplicate Rows")

print(df.duplicated().sum())
```

```python
# --------------------------------------------
# Step 13 : Verify Duplicate IDs
# --------------------------------------------

print("\nFinal Duplicate Order IDs")

print(df['OrderID'].duplicated().sum())
```

```python
# --------------------------------------------
# Step 14 : Dataset Summary
# --------------------------------------------

print("\nFinal Shape")

print(df.shape)

print("\nDataset Information")

print(df.info())
```

```python
# --------------------------------------------
# Step 15 : Export Cleaned Dataset
# --------------------------------------------

df.to_csv("Cleaned_Data_Analytics.csv", index=False)

df.to_excel(
    "Cleaned_Data_Analytics.xlsx",
    index=False
)

print("\nCleaned Dataset Saved Successfully")

print("CSV File : Cleaned_Data_Analytics.csv")

print("Excel File : Cleaned_Data_Analytics.xlsx")
```
