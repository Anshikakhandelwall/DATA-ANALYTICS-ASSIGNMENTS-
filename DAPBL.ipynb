**DATA CLEANING AND PROCESSING**

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
from scipy import stats

# Display missing values
print("Missing Values Before Treatment:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
df = df.drop_duplicates()
```

```python
# Handling Inconsistent and Noisy Data

# Remove extra spaces from text columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# Convert text columns to consistent case
if 'Department' in df.columns:
    df['Department'] = df['Department'].str.upper()

```

```python
# Outlier Detection Using IQR Method


if 'Percentage' in df.columns:
    Q1 = df['Percentage'].quantile(0.25)
    Q3 = df['Percentage'].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers_iqr = df[(df['Percentage'] < lower_bound) | (df['Percentage'] > upper_bound)]

    print("Outliers Detected Using IQR:")
    print(outliers_iqr)

    # Remove outliers
    df = df[(df['Percentage'] >= lower_bound) & (df['Percentage'] <= upper_bound)]
```

```python
# Outlier Detection Using Z-Score

if 'Marks_Obtained' in df.columns:
    z_scores = np.abs(stats.zscore(df['Marks_Obtained']))

    outliers_z = df[z_scores > 3]

    print("Outliers Detected Using Z-Score:")
    print(outliers_z)
```

```python
# Standardization
standard_scaler = StandardScaler()

scaled_standard = standard_scaler.fit_transform(
    df[['Marks_Obtained', 'Percentage', 'Attendance_Percentage']]
)

print("Standardized Data:")
print(scaled_standard[:5])

# Normalization
minmax_scaler = MinMaxScaler()

scaled_minmax = minmax_scaler.fit_transform(
    df[['Marks_Obtained', 'Percentage', 'Attendance_Percentage']]
)

print("Normalized Data:")
print(scaled_minmax[:5])
```

```python
cleaned_file = 'cleaned_student_performance.csv'
df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully.")
```

```python

# Hypothesis Testing - Chi-Square Test


# Example: Relationship between Department and Result_Status
contingency_table = pd.crosstab(df['Department'], df['Result_Status'])

chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

print("\nChi-Square Test Result:")
print("Chi-Square Statistic:", chi2)
print("P-Value:", p)

if p < 0.05:
    print("Department and Result Status are related.")
else:
    print("Department and Result Status are independent.")

```
