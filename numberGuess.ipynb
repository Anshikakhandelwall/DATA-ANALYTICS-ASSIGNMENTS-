RANDOM NUMBER GUESSING GAME

```python
import random
import time

# High scores for single player (session-based)
high_scores = {"Easy": 0, "Medium": 0, "Hard": 0}

print("🎮 Welcome to the Ultimate Number Guessing Game!")

# ================= GAME TYPE =================
print("\nChoose Game Type:")
print("1. Single Player")
print("2. Multiplayer")

game_type = int(input("Enter choice (1/2): "))

if game_type == 1:
    player_name = input("Enter your name: ")
else:
    player1 = input("Enter Player 1 name: ")
    player2 = input("Enter Player 2 name: ")
    players = [player1, player2]

    # Multiplayer stats
    player_stats = {
        player1: {"wins": 0, "score": 0},
        player2: {"wins": 0, "score": 0}
    }

# ================= GAME LOOP =================
while True:

    print("\nChoose Game Mode:")
    print("1. Normal Mode")
    print("2. Sudden Death (1 wrong guess = Game Over)")
    mode = int(input("Enter choice (1/2): "))

    print("\nChoose Difficulty Level:")
    print("1. Easy")
    print("2. Medium")
    print("3. Hard")
    choice = int(input("Enter your choice (1/2/3): "))

    if choice == 1:
        max_range, max_attempts, level = 20, 7, "Easy"
    elif choice == 2:
        max_range, max_attempts, level = 50, 6, "Medium"
    else:
        max_range, max_attempts, level = 100, 5, "Hard"

    if mode == 2:
        max_attempts = 1
        print("💀 Sudden Death Mode Activated!")

    secret_number = random.randint(1, max_range)
    attempts = 0
    guessed_numbers = set()
    hints_used = 0
    start_time = time.time()

    print(f"\n🔹 Level: {level}")
    print(f"🔹 Guess a number between 1 and {max_range}")
    print(f"🔹 Attempts Allowed: {max_attempts}")

    # ================= SINGLE PLAYER =================
    if game_type == 1:
        print(f"🔹 Player: {player_name}\n")

        while attempts < max_attempts:
            try:
                guess = int(input("Enter your guess: "))
            except ValueError:
                print("❌ Invalid input!")
                continue

            if guess in guessed_numbers:
                print("⚠️ Already guessed!")
                continue

            guessed_numbers.add(guess)
            attempts += 1

            if guess == secret_number:
                time_taken = int(time.time() - start_time)
                score = (max_attempts - attempts + 1) * 10
                bonus = max(0, 50 - time_taken)
                score += bonus

                print(f"\n🎉 Correct! Attempts: {attempts}")
                print(f"🏆 Score: {score} (Time Bonus: {bonus})")

                if attempts == 1:
                    print("🏅 Achievement: First Try Genius!")
                if time_taken <= 10:
                    print("⚡ Achievement: Speed Runner!")
                if hints_used == 0:
                    print("🎯 Achievement: No-Hint Master!")

                if score > high_scores[level]:
                    high_scores[level] = score
                    print("🔥 NEW HIGH SCORE!")

                break

            elif guess < secret_number:
                print("⬆️ Too Low!")
            else:
                print("⬇️ Too High!")

            if attempts == 3:
                hints_used += 1
                print("💡 Hint:", "EVEN" if secret_number % 2 == 0 else "ODD")

            if attempts == 4:
                hints_used += 1
                print(f"💡 Hint: Between {max(1, secret_number-10)} and {min(max_range, secret_number+10)}")

            if mode == 2:
                break

        if guess != secret_number:
            print(f"\n❌ Game Over! Correct number was {secret_number}")

        print("\n📊 Game Statistics")
        print(f"Attempts Used: {attempts}")
        print(f"Hints Used: {hints_used}")
        print(f"Numbers Guessed: {sorted(guessed_numbers)}")
        print(f"High Score ({level}): {high_scores[level]}")

    # ================= MULTIPLAYER =================
    else:
        print("\n🔹 Multiplayer Mode Started!")
        current_player = 0
        winner = None

        while attempts < max_attempts:
            player = players[current_player]
            print(f"\n🎯 {player}'s turn")

            try:
                guess = int(input("Enter your guess: "))
            except ValueError:
                print("❌ Invalid input!")
                continue

            attempts += 1

            if guess == secret_number:
                remaining = max_attempts - attempts
                points = 20 + remaining * 5

                player_stats[player]["wins"] += 1
                player_stats[player]["score"] += points
                winner = player

                print(f"\n🎉 {player} WINS THIS ROUND!")
                print(f"🏆 Points Earned: {points}")
                break

            elif guess < secret_number:
                print("⬆️ Too Low!")
            else:
                print("⬇️ Too High!")

            current_player = (current_player + 1) % 2

            if mode == 2:
                break

        if winner is None:
            print(f"\n❌ No one guessed the number!")
            print(f"The correct number was {secret_number}")

        print("\n📊 Multiplayer Scoreboard")
        for p, stats in player_stats.items():
            print(f"{p} → Wins: {stats['wins']} | Score: {stats['score']}")

    # ================= PLAY AGAIN =================
    play_again = input("\nPlay again? (y/n): ").lower()
    if play_again != 'y':
        print("👋 Thanks for playing!")
        break
```
