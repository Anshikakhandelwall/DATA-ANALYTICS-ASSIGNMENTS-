**Case Study 1: Student Marks Analysis**
Dataset:
marks = np.array([
[80, 75, 90],
[60, 88, 70],
[95, 92, 85],
[70, 65, 60],
[88, 78, 84]
])
**Tasks: **
1. Compute total marks of each student.
2. Calculate average marks per student.
3. Iden fy topper using NumPy.
4. Find subject-wise highest score.
5. Compute standard devia on of each subject.
6. Comment on performance consistency.

```python
import numpy as np

# Dataset (5 students × 3 subjects)
marks = np.array([
    [80, 75, 90],
    [60, 88, 70],
    [95, 92, 85],
    [70, 65, 60],
    [88, 78, 84]
])

# 1. Total marks of each student
total_marks = np.sum(marks, axis=1)
print("Total Marks of Each Student:", total_marks)

# 2. Average marks per student
average_marks = np.mean(marks, axis=1)
print("Average Marks of Each Student:", average_marks)

# 3. Identify topper (student with highest total)
topper_index = np.argmax(total_marks)
print("Topper is Student:", topper_index + 1)
print("Topper Total Marks:", total_marks[topper_index])

# 4. Subject-wise highest score
subject_highest = np.max(marks, axis=0)
print("Subject-wise Highest Scores:", subject_highest)

# 5. Standard deviation of each subject
subject_std = np.std(marks, axis=0)
print("Standard Deviation of Each Subject:", subject_std)

```

6.**Comment on Performance Consistency**

Subject 2 has the lowest standard deviation (~9.56).
 This means students performed more consistently in Subject 2.

Subject 1 has the highest deviation (~12.79).
 Marks are more spread out here (greater performance variation).

Student 3 shows consistently high performance across all subjects, making him the clear topper.

**Case Study 2: Temperature Analysis **
Dataset: temperature = np.array([32, 35, 31, 30, 29, 36, 34])
**Tasks: **
1. Compute mean temperature.
2. Find maximum and minimum temperature.
3. Calculate temperature variance.
4. Iden fy days above average temperature.
5. Normalize temperature values.
6. Interpret weather trend.

```python
import numpy as np

# Dataset
temperature = np.array([32, 35, 31, 30, 29, 36, 34])

# 1. Mean temperature
mean_temp = np.mean(temperature)
print("Mean Temperature:", mean_temp)

# 2. Maximum and Minimum temperature
max_temp = np.max(temperature)
min_temp = np.min(temperature)
print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)

# 3. Temperature Variance
variance_temp = np.var(temperature)
print("Temperature Variance:", variance_temp)

# 4. Days above average temperature
above_avg_days = np.where(temperature > mean_temp)
print("Indices of Days Above Average Temperature:", above_avg_days)
print("Temperatures Above Average:", temperature[above_avg_days])

# 5. Normalize temperature values (Min-Max Normalization)
normalized_temp = (temperature - min_temp) / (max_temp - min_temp)
print("Normalized Temperature Values:", normalized_temp)

```

6.**Interpretation of Weather Trend**

The temperatures fluctuate slightly but remain in a moderately warm range (29°C–36°C).

The week shows a dip in the middle (29–30°C) followed by a rise toward the end (34–36°C).

Overall, the weather trend indicates a slight warming pattern toward the end of the week.

Since variance is low (~6.53), the temperature changes are not extreme.
